```python
import numpy as np
import torch
import pandas as pd
from torch.utils.data import DataLoader
import os, sys
sys.path.append(os.path.abspath('../models'))
from Dataset import ProteinCSVWindowDataset, ProteinEmbeddingWithLabelsDataset,  collate_fn_window
from models.original import OriginalModel, OriginalModelLarger
from models.small import SmallModel

CHECKPOINT_PATH = './checkpoints/original_bs8_lr3e-04_ep15_20250623_232208.pth'
TEST_DATA_PATH = r"E:\Data\embeddings41k\test\all_embeddings_and_labels.h5"
OUTPUT_CSV = 'predictions_emb_23.csv'
BATCH_SIZE = 32
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'

# Modell und Testdaten laden
model = OriginalModel(input_size=1024, channels=512, num_classes=58)
model.load_state_dict(torch.load(CHECKPOINT_PATH, map_location=DEVICE))
model.to(DEVICE)
model.eval()

test_dataset = ProteinEmbeddingWithLabelsDataset(TEST_DATA_PATH)

test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_fn_window)
```

```python
results = []

with torch.no_grad():
    for batch in test_loader:
        inputs = batch["embeddings"]
        targets = batch["labels"]
        inputs = inputs.to(DEVICE)
        outputs, _ = model(inputs)
        preds = outputs.argmax(dim=-1).cpu().numpy()
        targets = targets.cpu().numpy()
        break
```

```python
import matplotlib.pyplot as plt
import os

os.makedirs('plots', exist_ok=True)

for x in range(new_targets.shape[0]):
    plt.figure(figsize=(12, 4))
    plt.plot(new_targets[x], label='Target', marker='o')
    plt.plot(preds[x], label='Prediction', marker='x')
    plt.title(f'Vergleich Target vs. Prediction für Sample {x}')
    plt.xlabel('Position (y)')
    plt.ylabel('Klassenindex (z)')
    plt.legend()
    plt.tight_layout()
    plt.savefig(f'plots/vergleich_{x}.png')
    plt.close()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import os

os.makedirs('plots', exist_ok=True)

num_classes = max(new_targets.max(), preds.max()) + 1

for x in range(new_targets.shape[0]):
    plt.figure(figsize=(12, 4))
    for arr, name in [(new_targets, 'Target'), (preds, 'Prediction')]:
        for cls in range(1, num_classes):  # Klasse 0 wird ausgelassen
            mask = (arr[x] == cls).astype(int)
            plt.plot(mask, label=f'{name} Klasse {cls}')
    plt.title(f'Klassen-Visualisierung für Sample {x}')
    plt.xlabel('Position (y)')
    plt.ylabel('Klassen-Präsenz')
    plt.legend()
    plt.tight_layout()
    plt.savefig(f'plots/klassen_visualisierung_{x}.png')
    plt.close()
```
